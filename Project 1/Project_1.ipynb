```python
import pandas as pd
import numpy as np

df = pd.read_csv("Dataset for Data Analytics - Sheet1.csv")
```

```python
df.shape
```

```python
numeric_cols = ["Quantity", "UnitPrice", "ItemsInCart", "TotalPrice"]

Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
outliers_totalprice = df[
    (df["TotalPrice"] < lower_bound["TotalPrice"]) |
    (df["TotalPrice"] > upper_bound["TotalPrice"])
]

outliers_totalprice
```

```python
df["CouponCode"] = df["CouponCode"].fillna("No Coupon")
```

```python
df["CouponCode"].isnull().sum()
```

```python
df["TotalPrice"]=df["TotalPrice"].clip(upper_bound["TotalPrice"])
```

```python
df["TotalPrice"].max()
```

```python
upper_bound["TotalPrice"]
```

```python
df["TotalPrice"] = df["TotalPrice"].clip(upper=upper_bound["TotalPrice"])
```

```python
df["TotalPrice"].max()
```

```python
df["HasCoupon"] = (df["CouponCode"] != "No Coupon").astype(int)
```

```python
df[["CouponCode", "HasCoupon"]].head(10)
```

```python
df["PricePerItem"] = df["TotalPrice"] / df["Quantity"]
```

```python
df[["TotalPrice", "Quantity", "PricePerItem"]].head()
```

```python
df["CartUtilization"] = df["Quantity"] / df["ItemsInCart"]
```

```python
df[["Quantity", "ItemsInCart", "CartUtilization"]].head()
```
